## 7. Create a dataframe with missing values.


1. Drop the row with the missing value in column `c`.
2. Instead of dropping the row with the missing value, replace the missing value with `-99`.
3. Instead of replacing the missing value with `-99`, fill it with the median of the corresponding column.

````python
df_missing = pd.DataFrame(
    {"a": [1, 2, 3], "b": [4, 10, 6], "c": [7, np.nan, 9]}
)

# 1. drop the row with the missing value in column "c"
df_missing.dropna(subset = ["c"])

# 2. replace the missing value with -99
df_missing.replace(np.nan, -99)

# 3. fill missing values by the median of a column
median = df_missing["c"].median()
df_missing.fillna({'c': median}, inplace=True)
df_missing
````

## 10. Create a dataframe and calculate the mean, max, min, and counts by animal type (Falcon and Parrot). Hint: use `.groupby()`.

````python
animal_speed = pd.DataFrame(
    {"Animal": ["Falcon", "Falcon", "Falcon", "Parrot", "Parrot", "Parrot"],
     "Max Speed": [380, 370, 360, 24, 26, 25]}
)

animal_speed.groupby(by = "Animal").mean()
animal_speed.groupby(by = "Animal").max()
animal_speed.groupby(by = "Animal").min()
animal_speed.groupby(by = "Animal").count()
````

## 11. Create a dataframe.

## 12. Create two dataframes.

1. Left merge `df1` with `df2` by `x1`.
2. Inner merge `df1` with `df2` by `x1`.
3. Outer merge `df1` with `df2` by `x1`.

````python
df1 = pd.DataFrame(
    {"x1": ["a", "b", "c"], "x2": [10, 20, 30]}
)

df2 = pd.DataFrame(
    {"x1": ["b", "c", "d"], "x3": ["beta", "cycle", "tibble"]}
)

# left
pd.merge(df1, df2, how = "left", on = "x1")

# inner
pd.merge(df1, df2, how = "inner", on = "x1")

# outer
pd.merge(df1, df2, how = "outer", on = "x1")
````
